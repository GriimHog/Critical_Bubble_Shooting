## 1. The potential

$V(\phi,T) = D(T^2-T_0^2)\phi^2 - E\,T\,\phi^3 + \tfrac{\lambda}{4}\phi^4$

```python
lam, D, E, T0 = 0.1, 0.1, 0.05, 1.0

def V(phi, T):
    return (D * (T ** 2 - T0 ** 2) * phi ** 2 - E * T * phi ** 3 + (lam / 4) * phi ** 4)
    
def dV(phi, T):
    return (D * (T ** 2 - T0 ** 2) * phi * 2 - 3 * E * T * phi ** 2 + lam * phi ** 3)
```

### Check

```python
h = 1e-6
phi, T = 0.8, 1.1
print(dV(phi, T), (V(phi + h, T) - V(phi - h, T)) / (2 * h))
```

```python
def disc2(T):
    a, b, c = lam, (-3 * E * T), 2 * D * (T ** 2 - T0 ** 2)
    return b**2 - 4 * a * c
    
def extrema(T):
    a, b, c = lam, (-3 * E * T), 2 * D * (T ** 2 - T0 ** 2) # solve lam*phi^2 - 3*E*T*phi + 2*D*(T^2-T0^2) = 0 with the quadratic formula.
    disc = b**2 - 4 * a * c
    if disc >= 0:
        phi_m, phi_p = ((-b - math.sqrt(disc)) / (2 * a)), ((-b + math.sqrt(disc)) / (2 * a))
        return phi_m, phi_p
    else:
        return None

# Analytic critical values from your derivation
Tc = math.sqrt(T0**2 / (1 - (E ** 2 / (lam * D))))
phic = 2 * E * Tc / lam
sigma = math.sqrt(lam/2) * phic**3 / 6
print(Tc, phic, sigma)
```

## 2. The integrator (Step 3)

Equation: $\phi'' + \tfrac{2}{r}\phi' = V'(\phi)$. Write it as a first-order system $y=(\phi,\phi')$:
$y_0' = y_1$ and $y_1' = V'(y_0) - \tfrac{2}{r}y_1$.

The false vacuum is at $\phi = 0$. Two events stop the integration:
- **overshoot**: $\phi$ crosses 0 going downward (event function `y[0]`, direction -1),
- **undershoot**: $\phi'$ becomes positive (event function `y[1]`, direction +1).

```python
def shoot(phi0, T, rmax=200.0, r0=1e-4, rtol=1e-10, atol=1e-12):
    """Integrate from r0 with initial value phi0.
    Return (flag, sol): flag = +1 overshoot, -1 undershoot, 0 neither before rmax."""
    c = dV(phi0, T)
    # TODO: small-r start from your derivation:
    #   phi(r0)  = phi0 + c*r0**2/6
    #   phi'(r0) = c*r0/3
    y0 = None

    def rhs(r, y):
        # TODO: return [y[0]', y[1]'] as above
        pass

    def over(r, y):  return y[0]
    over.terminal, over.direction = True, -1
    def under(r, y): return y[1]
    under.terminal, under.direction = True, 1

    sol = solve_ivp(rhs, [r0, rmax], y0, events=[over, under],
                    rtol=rtol, atol=atol, dense_output=True)
    if sol.t_events[0].size: return +1, sol
    if sol.t_events[1].size: return -1, sol
    return 0, sol
```

## 3. Bisection on phi0 (Step 4)

Overshoot means phi0 is too close to the broken minimum, so lower `hi`. Undershoot means raise `lo`.

```python
def bounce(T, tol=1e-12):
    pm, pp = extrema(T)
    lo = pm + 1e-9*(pp - pm)
    hi = pp - 1e-9*(pp - pm)
    # TODO: loop (at most 200 times):
    #   mid = (lo+hi)/2, flag, _ = shoot(mid, T)
    #   flag > 0 -> hi = mid, otherwise lo = mid
    #   stop when hi - lo < tol * pp
    phi0 = None
    flag, sol = shoot(phi0, T)
    return phi0, sol
```

**Warning for (c).** The bubble radius is about $2\sigma/\varepsilon$. Near Tc it becomes so large that the shooting
cannot follow it in double precision (the solution is exponentially sensitive to phi0). For T above about 1.150 your S3 will
silently come out too small, because the integration stopped before the bubble wall. Always compare the integration end
`sol.t[-1]` with $2\sigma/\varepsilon$. If the end is smaller than that, the result is not valid.
